```python
df.isnull().sum() #To check how many null values are there and where
```

```python
df['Age'].fillna(df['Age'].mean(), inplace=True) #To fill all null values in age  with mean of age 
```

```python
df['Embarked'].fillna(df['Embarked'].mode, inplace=True) # To fill all null values in embarked with most comman values in embarked
```

```python
df.drop(columns =['Cabin'] , inplace =True) # Drop the cabin column because more than 70% null values in cabin column
```

```python
df.isnull().sum() #After fill and drop here is clean data with 0 null values
```

```python
df.duplicated().sum() # Check how many duplicate value in dataset
```

```python
df.drop_duplicates(inplace=True) #Drop all duplicate values in dataset
```

```python
df.duplicated().sum()  #After drop a duplicate here is a 0 duplicate value
```

```python
df['Survived']= df['Survived'].astype('bool')
df['Pclass'] = df['Pclass'].astype('category')
df['Sex'] = df['Sex'].astype('category')
df['Embarked'] = df['Embarked'].astype('category')
```

```python
df.dtypes #After change the data type here is all correct data types
```

```python
df.rename(columns={
    'PassengerId' : 'Passenger_Id',
    'Pclass' : 'Passenger_class',
    'SibSp' : 'Siblings_Spouses',
    'Parch' : 'Parents_Children',
    'Embarked': 'Port_of_Embarkation'
}, inplace=True)  # Change the columns name for better understanding
```
